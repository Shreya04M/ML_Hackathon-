# PathSnap dataset exploration

Read-only exploratory analysis of the raw dataset. This notebook does not clean data, generate candidates, extract features, or train models.

```python
from pathlib import Path
from dotenv import load_dotenv

from preprocessing.config import get_dataset_root
from preprocessing.data_exploration import explore_dataset, markdown_report

dataset_root = get_dataset_root()
print(f'Dataset root: {dataset_root}')
print(f'Exists and accessible: {dataset_root.is_dir()}')
```

```python
summary = explore_dataset(dataset_root)
for partition, details in summary.get('partitions', {}).items():
    print(partition, details['trip_count'], 'trips')
    print('file counts:', dict(details['file_counts']))
```

```python
for partition, details in summary.get('partitions', {}).items():
    print(f'\n[{partition}] representative: {details.get("representative", {}).get("trip_id")}')
    for kind, info in details.get('representative', {}).get('files', {}).items():
        print(kind, 'rows=', info['rows'], 'columns=', info['columns'], 'header=', info['header_detected'], 'missing=', info['missing_values'])
        print('sample:', info['sample'])
```

```python
print('Basic numeric ranges for representative files:')
for partition, details in summary.get('partitions', {}).items():
    for kind, info in details.get('representative', {}).get('files', {}).items():
        print(partition, kind, 'min=', info['numeric_min'], 'max=', info['numeric_max'])
```

## Initial observations

The report below records measured inventory and representative file summaries. File identifiers are preserved as strings. Track observations and route arcs are not assumed to correspond row-for-row.

```python
report_path = Path('experiments/results/dataset_exploration_report.md')
report_path.parent.mkdir(parents=True, exist_ok=True)
report_path.write_text(markdown_report(summary), encoding='utf-8')
print(f'Wrote {report_path}')
```
